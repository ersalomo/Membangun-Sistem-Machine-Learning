# **1. Perkenalan Dataset**

Dataset yang digunakan dalam eksperimen ini adalah **Wine Quality Dataset** versi modifikasi (sintetis) untuk keperluan simulasi MLOps. Dataset ini mengandung berbagai parameter kimia dari anggur (*wine*) beserta label kualitasnya.

**Fitur-fitur dalam dataset:**
- `fixed_acidity`: Keasaman tetap anggur.
- `volatile_acidity`: Keasaman mudah menguap.
- `citric_acid`: Kandungan asam sitrat (memiliki beberapa nilai kosong/NaN untuk simulasi).
- `residual_sugar`: Sisa kadar gula setelah fermentasi.
- `chlorides`: Kandungan garam/klorida.
- `free_sulfur_dioxide`: Kandungan sulfur dioksida bebas.
- `total_sulfur_dioxide`: Kandungan total sulfur dioksida.
- `density`: Kepadatan anggur.
- `pH`: Tingkat keasaman (memiliki beberapa nilai kosong/NaN untuk simulasi).
- `sulphates`: Kandungan sulfat.
- `alcohol`: Kadar alkohol.
- **Target (`quality`)**: Label biner (0 untuk kualitas standar/rendah, 1 untuk kualitas tinggi).

# **5. Data Preprocessing**

Berdasarkan temuan EDA di atas, langkah preprocessing wajib yang kita lakukan meliputi:
1. **Menghapus Data Duplikat** untuk menjaga integritas data.
2. **Menangani Missing Values**: Mengisi nilai kosong pada kolom `citric_acid` dan `pH` menggunakan nilai median kolom masing-masing.
3. **Normalisasi/Standarisasi Fitur**: Menggunakan `StandardScaler` untuk menyamakan skala fitur numerik sebelum model dilatih.
4. **Menyimpan Dataset Hasil Preprocessing** ke dalam file siap pakai.

```python
# 1. Menghapus data duplikat
df_clean = df.drop_duplicates().copy()
print(f"Banyak data setelah menghapus duplikat: {df_clean.shape[0]} (sebelumnya: {df.shape[0]})")
```

```python
# 2. Menangani missing values dengan median
impute_cols = ['citric_acid', 'pH']
for col in impute_cols:
    median_val = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(median_val)
    print(f"Mengisi nilai kosong pada '{col}' dengan nilai median: {median_val:.4f}")

print("Jumlah nilai kosong setelah pembersihan:")
print(df_clean.isnull().sum())
```

```python
# 3. Standarisasi Fitur Numerik
features = df_clean.drop(columns=['quality'])
target = df_clean['quality']

scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

# Konversi kembali ke DataFrame
df_preprocessed = pd.DataFrame(scaled_features, columns=features.columns)
df_preprocessed['quality'] = target.reset_index(drop=True)

print("Dimensi data hasil preprocessing:", df_preprocessed.shape)
df_preprocessed.head()
```

```python
# 4. Menyimpan dataset hasil preprocessing
output_dir = "./namadataset_preprocessing"
os.makedirs(output_dir, exist_ok=True)

output_path = os.path.join(output_dir, "wine_preprocessed.csv")
df_preprocessed.to_csv(output_path, index=False)

# Simpan juga langsung ke level preprocessing untuk kemudahan akses model
df_preprocessed.to_csv("./wine_preprocessed.csv", index=False)

print(f"Dataset preprocessing berhasil disimpan ke {output_path}!")
```
